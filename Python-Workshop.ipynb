```python
def change_into_datatime(col):
  df[col]=pd.to_datetime(df[col])
```

# Data visualisation

```python
t=1
plt.figure(figsize=(15,10))
for i in num_col:
  plt.subplot(4,3,t)
  sns.distplot(df[i])
  plt.title(f'{i} skewness:{round(df[i].skew(),2)}')
  t+=1
plt.tight_layout()
plt.show()
```
