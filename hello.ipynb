### Union

```python
A={1,2,3}
B={3,4,5}
result =A.union(B)
print(result)
print(A.intersection(B))
print(A|B)
print(A&B)
```

### Dictionary

```python
# Duplicates not allowed
thisdict = {
    "brand":"ford",
    "model":"Mustang",
    "year": 1964,
    "year":2020
}
print(thisdict)
print(len(thisdict))
```
