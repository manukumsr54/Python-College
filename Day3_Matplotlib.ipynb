# what is Matplotlib ?

Matplotlib is a python library used to create : Line Charts , Bar charts , Pie charts and scatter plots , Histogram plots area charts Heatmaps(with other libraries) it helps you visualize data , making patterns and trends easier to understand

### Line Chart

```python
import matplotlib.pyplot as plt

months = ["Jan","Feb","Mar","Apr","May"]
sales = [100,150,120,180,200] 

plt.plot(months,sales)
plt.ylabel("Sales")
plt.xlabel("Monthly")
plt.show()
```

### Change Line Style

```python
plt.plot(
    months,
    sales,
    color="red",
    linestyle="--",
    linewidth=2,
    marker="o"
)
```

```python
months = ["Jan","Feb","Mar","Apr","May"]
sales_2025 = [100,150,120,180,220] 
sales_2026 = [120,170,160,200,250]

plt.plot(months,sales_2025,marker="o")
plt.plot(months,sales_2026,marker="o")
plt.title("Monthly sales")
plt.ylabel("Sales")
plt.xlabel("Monthly")
plt.show()
```

```python
import matplotlib.pyplot as plt

students = ["vicky","Rahul","Aman","Neha"]
marks = [85,90,78,95]

plt.bar(students,marks,width=0.6)
plt.title("Student Marks")
plt.ylabel("Marks")
plt.xlabel("Student")
plt.show()
```

```python
# Horizontal
plt.barh(students,marks)
plt.show()
```

```python
# Multiple Bar Charts
students = ["A","B","C","D"]
python = [80,70,90,60]
java = [70,75,85,80]

import numpy as np
x = np.arange(len(students))
width = 0.35
plt.bar(x-width/2,python,width,label="python")
plt.bar(x+width/2,java,width,label="java")
plt.xticks(x,students)
plt.legend()
plt.show()
```

### Pie Charts

```python
import matplotlib.pyplot as plt

subject = ["AI", "Python", "Java", "SQL"]
hours = [23, 53,35,20]
explode = [0.1, 0, 0, 0]
mycolors = ["red", "blue", "green", "yellow"]

plt.pie(hours, 
        labels=subject,
        explode=explode,
    autopct="%1.1f%%",
    shadow=True,
    colors = mycolors
        )
plt.title("study time")
plt.legend()
plt.show()
```
